```python
ROOT = Path.cwd()

TEMPLATE_PATH = ROOT / "dashboard_template.html"
OUTPUT_PATH = ROOT / "dashboard.html"

print("Template:", TEMPLATE_PATH)
print("Output:", OUTPUT_PATH)
```

## Template Helpers

These functions generate the dynamic pieces that will be inserted into the HTML template.

```python
def make_hour_options() -> str:
    return "\n".join(
        f'              <option value="{hour}">{hour:02d}:00</option>'
        for hour in range(24)
    )
```

```python
def make_dashboard_config() -> str:
    config = {
        "dataRoot": "data",
        "manifestPath": "data/flow_filter_json/flow_filter_manifest.json",
        "geojson": {
            "gu": "data/seoul_gu.geojson",
            "dong": "data/seoul_dong.geojson",
            "station": "data/station.geojson",
        },
        "geoIdFields": {
            "gu": "ADM_NM",
            "dong": "ADM_NM",
            "station": "stationID",
        },
        "defaultFilters": {
            "level": "gu",
            "month": "all",
            "weekday_type": "all",
            "hour": "all",
            "flowPercent": 5,
        },
    }
    return f"<script>\n  window.DASHBOARD_CONFIG = {json.dumps(config, indent=2)};\n</script>"
```

## Render Dashboard

This cell reads the template and replaces each placeholder token with dashboard content.

```python
def render_dashboard() -> str:
    template = TEMPLATE_PATH.read_text(encoding="utf-8")
    replacements = {
        "{{PAGE_TITLE}}": "Seoul Bike-Sharing Flow Dashboard",
        "{{DASHBOARD_TITLE}}": "Seoul Bike-Sharing Flow Dashboard <br>(January & August, 2025)",
        "{{DASHBOARD_SUBTITLE}}": (
            "Interactive origin-destination flows by month, day type, hour, "
            "and spatial level."
        ),
        "{{HEADER_BADGE}}": "Interactive Geovisualization",
        "{{HOUR_OPTIONS}}": make_hour_options(),
        "{{JS_DASHBOARD_CONFIG}}": make_dashboard_config(),
        "{{FOOTER_TEXT}}": 'Data source: <a href="https://data.seoul.go.kr">Seoul Open Data Plaza</a>',
    }

    html = template
    for token, value in replacements.items():
        html = html.replace(token, value)
    return html
```

## Build Final HTML

Run this cell whenever you edit `dashboard_template.html`, `style.css`, or the dashboard configuration.

```python
html = render_dashboard()
OUTPUT_PATH.write_text(html, encoding="utf-8")

print(f"Wrote {OUTPUT_PATH}")
```

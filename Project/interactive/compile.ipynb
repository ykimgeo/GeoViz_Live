# Compile the dashboard template into `dashboard.html`

This notebook reads Python-prepared JSON data, injects it into the HTML template as `window.CHART_DATA`, and writes the final browser-ready dashboard file.

```python
from pathlib import Path
import json

# Resolve all paths relative to the interactive dashboard folder.
project_dir = Path.cwd()
template_path = project_dir / "dashboard_template.html"
chart_data_path = project_dir / "data" / "chart_data.json"
output_path = project_dir / "dashboard.html"

template_path, chart_data_path, output_path
```

```python
# Read the HTML template and the processed JSON payload.
template_html = template_path.read_text(encoding="utf-8")
chart_data = json.loads(chart_data_path.read_text(encoding="utf-8"))

# Pretty-print the JSON so the compiled HTML stays readable.
chart_data_json = json.dumps(chart_data, ensure_ascii=False, indent=2)

# Wrap the JSON in a script tag so JavaScript can read window.CHART_DATA.
js_chart_data = "<script>\nwindow.CHART_DATA = " + chart_data_json + ";\n</script>"

chart_data["meta"]
```

```python
# Replace template placeholders with values prepared in Python.
compiled_html = template_html
compiled_html = compiled_html.replace("{{PAGE_TITLE}}", chart_data["meta"].get("pageTitle", "Interactive Dashboard"))
compiled_html = compiled_html.replace("{{DASHBOARD_TITLE}}", chart_data["meta"].get("dashboardTitle", "Interactive Dashboard"))
compiled_html = compiled_html.replace("{{DASHBOARD_SUBTITLE}}", chart_data["meta"].get("dashboardSubtitle", ""))
compiled_html = compiled_html.replace("{{JS_CHART_DATA}}", js_chart_data)

# Write the final standalone dashboard file.
output_path.write_text(compiled_html, encoding="utf-8")
print(f"Compiled dashboard written to: {output_path}")
```
